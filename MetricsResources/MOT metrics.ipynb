### **MOT + MOTS  metrics**

We will discuss different metrics used in multiple object tracking in a given video sequence.


We take a look at available challenges and datasets.

[metrics+datasets+challenges](https://motchallenge.net/)




**Qualities of a GOOD Multiple Object TRACKER**

* It should at all points in time find the correct number of objects present and estimate the position of each object as precisely as possible. 
(note that properties such as the contour, orientation, or speed of objects are not explicitly considered here). 

* It should also keep consistent track of each
object over time: each object should be assigned a unique
track ID which stays constant throughout the sequence (even
after temporary occlusion, etc.)

**Method**

---

Assuming that for every time frame t, a multiple
object tracker outputs a set of hypotheses $\{h_1,...,h_m \}$ for a
set of visible objects $\{ o_1,...,o_n \}$ , the evaluation procedure
comprises the following steps:


**For each time frame t,**


    1. establish the best possible correspondence between hypotheses $h_j$ and objects $o_$ ,

    2. for each found correspondence, compute the error in the object’s position estimation,

    3. accumulate all correspondence errors:

        - count all objects for which no hypothesis was output as misses **TP**
          (IoU, Jaccard index...) if the distance between object and 
          is smaller than a given threshold it's considered a match.

        - count all tracker hypotheses for which no real object exists as **FP**

        - count all occurrences where the tracking hypothesis for an object
          changed compared to previous frames as mismatch errors. 
          This could happen, for example, 
          when two or more objects are swapped as they pass close to each other,
          or when an object track is reinitialized with a diﬀerent track ID,
          after it was previously lost because of occlusion. 
          **IDS (identity switches)**


**Hypothesis-Object Matching process**

    Let M  = {(oi,hj), ... } be a mapping of o and h at time t.
    For every time frame t, consider the following.

      1.  For every mapping (oi , hj ) in Mt−1 , 
          verify if it is still valid. If object o i is still visible 
          and tracker hypothesis h j still exists at time t, 
          and if their distance does not exceed the threshold T, 
          make the correspondence between o i and h j for frame t

      2. For all objects for which no correspondence was made yet, 
         try to find a matching hypothesis. one-to-one and pairs for 
         which the distance does not exceed T(threshold)

         If correspondence(oi,hk) is made that contradicts a mapping (oi,hj) in
         Mt−1 ,  replace (o i , h j ) with (o i , h k ) in Mt . Count this as
         mismatch error and let mmet be the # of mismatch errors of frame t.

      3. Let ct be the number of matches found for time t get distances d

      4. All remaining hypotheses are considered FP. Similarly, 
         all remaining objects are considered misses (FN). 
         let fpt: TP,  mt: misses ,  gt:objects present  at time 

      5. repeat the steps and accumulate over the video sequence
      

## **MOT**

---

**The multiple object tracking accuracy (MOTA)**


$MOTA=1 - \frac{\sum_{t}(m_t + fp_t + mme_t )} {\sum_{t} g_t}$

where $m_t , fp_t , mme_t,g_t$ are misses(FN) , false positives, mismatch error ,total object present at time $t$ respectively,


    MOTA can be seen as a combination of 3 individual ratios 
    ... see below


**The multiple object tracking precision (MOTP)**

---

$MOTP=\frac{\sum_{i,t} d_t^i} {\sum_{t} c_t}$

where $d_t^i$ is the distance between hypothesis and ground truth from M tuple,
$c_t$ is total matches count for time t


      It is the total error in estimated position for matched
      object-hypothesis pairs over all frames, 
      averaged by the total number of matches made.


**Misses Ratio $\bar{m}$ = $(FN ratio)$** $↓$

## **HOTA**

considers detection association and localization 

<pre>

<b>Matching predictions with ground truths</b>

Mathing occurs at detection level in HOTA

TP is a pair of prDet and gtDet where Similarity S >= α threshold 
FN is a gtDet that is not matched to any prDet
FP is a prDet that is not matched to any 

  The matching between gtDets and prDets is bijective (one-to-one) in each frame

  since multiple matches are possible,
  they use hungarian algorithm to find matches that would maximize HOTA


<b>Measuring Association</b>
to measure association : 

    TPAs (True Positive Associations)
    FNAs (False Negative Associations)
    FPAs (False Positive Associations)

are defined for each TP pair

for a given TP, c:

TPA(c) = {k},
k ∈ {TP | prID(k) = prID(c) ∧ gtID(k) = gtID(c)}


FNA (c) = {k},
    {TP | prID(k) != prID(c) ∧ gtID(k) = gtID(c)}
k ∈
    ∪ {FN | gtID(k) = gtID(c)}

FPA (c) = {k},
    {TP | prID(k) = prID(c) ∧ gtID(k) = gtID(c)}
k ∈
    ∪ {FP | prID(k) = prID(c)}

</pre>


$HOTA_α =\sqrt{\frac{\sum_{cϵ\{TP\}}{Α(c)}} {|TP| + |FN| + |FP|}}$

where $𝛢(c)$ is the association score of gtTraj and prTraj for a given TP c

$A(c) =\frac{TPA(c)} {|TPA(c)| + |FNA(c)| + |FPA(c)|}$


<pre>
<b>Matching to Optimise HOTA</b>

Hungarian algorithm is run to select matches that maximize:
  TP true positives
  A  mean association score 
  S  localization similarity

</pre>

$HOTA = \int_0^1{HOTA_{α} d{α}} ≈ \frac{1}{19} \sum_{α ϵ0.05:0.95:0.05} {HOTA_{α}}$


**DetRe**

\begin{align}
 DetRe_α=\frac{{|TP|}}{|TP|+|FN|}
\end{align}

```
Detection Recall. tracker fails to detect gtDets that exist
TP /(TP + FN) averaged over localization thresholds.
```

**DetPr**
\begin{align}
 DetPr_α=\frac{{|TP|}}{|TP|+|FP|}
\end{align}

```
Detection Precision : tracker predicts extra prDets that dont exit in gtDets
TP /(TP + FP) averaged over localization thresholds.
```

**DetA**

\begin{align}
 DetA_α=\frac{|TP|}{|TP| + |FN| + |FP|}
\end{align}

```
Detection Accuracy
Detection Jaccard index averaged over localization thresholds.
```


**AssRe**

\begin{align}
 AssRe_α=\frac{1}{|TP|} \sum_{cϵ\{TP\}}\frac{|TPA(c)|}{|TPA(c)| + |FNA(c)|}
\end{align}

```
Association Recall : tracker assigns diff prIDs to the same gtTraj
TPA / (TPA + FNA) averaged over all matching detections 
and then averaged over localization thresholds.
```

**AssPr**

\begin{align}
 AssPr_α=\frac{1}{|TP|} \sum_{cϵ\{TP\}}\frac{|TPA(c)|}{|TPA(c)| + |FPA(c)|}
\end{align}

```
Association Precision tracker assigns same prID to 2 diff gtTraj
TPA / (TPA + FPA) averaged over all matching detections 
and then averaged over localization thresholds.
```

**AssA**

\begin{align}
 AssA_α=\frac{AssRe_α . AssPr_α}{AssRe_α + AssPr_α - AssRe_α . AssPr_α}
\end{align}

```
Association Accuracy
Association Jaccard index averaged over all matching
detections and then averaged over localization thresholds.
```

**LocA**

\begin{align}
 LocA=\int_0^1{\frac{1}{|TP_α|}  \sum_{cϵ\{TP_α\}}{S(c)d_α}}
 \space where S(c) \space is \space spatial \space similarity \space score 
\end{align}

```
Localization Accuracy : gtDets and prDets are not spatially well aligned
Average localization similarity averaged over all
matching detections and averaged over localization thresholds.
```
**HOTA**
\begin{align*}
HOTA_α =\sqrt{\frac{\sum_{cϵ\{TP\}}{Α(c)}} {|TP| + |FN| + |FP|}}
=\sqrt{DetA_α . AssA_α}
\end{align*}


```
Higher Order Tracking Accuracy
Geometric mean of detection accuracy and
association accuracy. Averaged across localization thresholds.
```
